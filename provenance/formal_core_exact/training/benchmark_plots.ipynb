# Distribution of THDot

```python
with open("temp/thdots_crawler_10.npy", "rb") as f:
    thdots_crawler = np.load(f)
with open("temp/thdots_ring_10.npy", "rb") as f:
    thdots_ring = np.load(f)
```

```python
sns.kdeplot(thdots_crawler, bw=0.01, label="crawler")
sns.kdeplot(thdots_ring, bw=0.01, label="ring")
plt.xlabel("$\dot\\theta$")
plt.title("Distribution of values of $\dot\\theta$ for an episode with simple non-reciprocity")
plt.legend()
plt.savefig("final_plots/thdot_distribution.png",  bbox_inches='tight')
```

```python
with open("temp/obs_crawler_10.npy", "rb") as f:
    obs_crawler = np.load(f)
with open("temp/obs_ring_10.npy", "rb") as f:
    obs_ring = np.load(f)
```

```python
points = obs_crawler
x = points[:, 0]
y = points[:, 1]

# Create heatmap
plt.figure(figsize=(6, 5))
plt.hist2d(x, y, bins=100, cmap='hot', range=[[-np.pi, np.pi], [-np.pi, np.pi]],)

plt.colorbar(label='Count')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('2D Point Density Heatmap')
plt.show()
```

```python
points1 = obs_crawler
points2 = obs_ring
x1, y1 = points1[:, 0], points1[:, 1]
x2, y2 = points2[:, 0], points2[:, 1]

# Shared settings
range_xy = [[-np.pi, np.pi], [-np.pi, np.pi]]
bins = 100
cmap = 'inferno'

fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharex=True, sharey=True)

# Common color scale
H1, _, _ = np.histogram2d(x1, y1, bins=bins, range=range_xy)
H2, _, _ = np.histogram2d(x2, y2, bins=bins, range=range_xy)

# Compute global vmin/vmax based on both datasets
vmin = min(H1.min(), H2.min())
vmax = max(H1.max(), H2.max())

print(f"Auto-scaled vmin={vmin:.2f}, vmax={vmax:.2f}")

# Heatmaps
h1 = axes[0].hist2d(x1, y1, bins=bins, range=range_xy, cmap=cmap, vmin=vmin, vmax=vmax)
h2 = axes[1].hist2d(x2, y2, bins=bins, range=range_xy, cmap=cmap, vmin=vmin, vmax=vmax)

axes[0].set_title("Crawler")
axes[1].set_title("Ring")

for ax in axes:
    ax.set_xlabel("$\delta\\theta_{i+1}$")
axes[0].set_ylabel("$\delta\\theta_{i-1}$")

# Create a single colorbar axis on the right
cbar = fig.colorbar(h1[3], ax=axes, location='bottom', aspect=40)
pos = cbar.ax.get_position()          # get its [left, bottom, width, height]
new_pos = [pos.x0, pos.y0 - 0.35, pos.width, pos.height]
cbar.ax.set_position(new_pos)
cbar.set_label("Frequency")
plt.suptitle("Distribution of observation values for an episode with simple non-reciprocity")

plt.tight_layout()
# plt.savefig("final_plots/obs_dist.png",  bbox_inches='tight')
```
