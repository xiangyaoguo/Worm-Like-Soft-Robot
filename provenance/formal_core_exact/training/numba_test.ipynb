```python
import numpy as np
import numba as nb
from numba import njit, jit, int32
from types import MethodType
```

```python
def infer_numba_signature(*args):
    output = nb.typeof(args[0])
    inputs = [nb.typeof(a) for a in args[1:]]
    return output(*inputs)

a = True
def do_stuff(array):
    if a:
        return array + 1
    else:
        return array - 1
do_stuff_better = njit(infer_numba_signature(
    np.array([1j]),
    np.array([1j])
))(do_stuff)

def fuck_around(array):
    return do_stuff_better(array) + 1j
fuck_around_better = njit(infer_numba_signature(
    np.array([1j]),
    np.array([1j])
))(fuck_around)
```

```python
a = False
fuck_around_better(np.array([1j]))
```

```python
do_stuff_better(np.array([1]))
```

```python
class Operator:
    def __init__(self, add=1):
        def func(array):
            return array + add
        self.run = njit(infer_numba_signature(
            np.array([1]),
            np.array([1])
        ))(func)
```

```python
op1 = Operator(3)
op2 = Operator(-1)
op3 = Operator(10)
```

```python
print(op1.run(np.array([1])))
print(op2.run(np.array([1])))
print(op3.run(np.array([1])))
```
