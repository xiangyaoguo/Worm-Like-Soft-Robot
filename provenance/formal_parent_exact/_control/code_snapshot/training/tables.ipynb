```python
from experiment_sets import set_tunnel
from sim import TrainedPolicySim
from collections import defaultdict
```

```python
best_checkpoints = {
    "flat": set_tunnel.best_group_checkpoints["('ddpg', True, 'flat')"],
    "stairs": set_tunnel.best_group_checkpoints["('ddpg', False, 'stairs')"],
    "tunnel": set_tunnel.best_group_checkpoints["('ddpg', True, 'tunnel')"],
    "mix": set_tunnel.best_group_checkpoints["('ddpg', True, '[\"stairs\", \"tunnel\", \"flat\"]')"]
}
sims = {k: TrainedPolicySim(f"results/{v[0]}/checkpoint_{v[1]}.pt", num_envs=100) for k, v in best_checkpoints.items()}
```

```python
best_checkpoints
```

```python
envs = {k: v.env for k, v in sims.items()}
```

```python
scores = defaultdict(dict)
for trained in ["flat", "stairs", "tunnel", "mix"]:
    sim = sims[trained]
    for evaluate in ["flat", "stairs", "tunnel"]:
        sim.env = envs[evaluate]
        scores[trained][evaluate] = sim.evaluate_speed()
```

```python
for trained in ["flat", "stairs", "tunnel", "mix"]:
    sc = scores[trained]
    mean = sum([s[0] for s in sc.values()]) / len(sc)
    print(trained, "&", "{:.2f}".format(sc["flat"][0]*100), "&", "{:.2f}".format(sc["stairs"][0]*100), "&", "{:.2f}".format(sc["tunnel"][0]*100),
         "&", "{:.2f}".format(mean*100), "\\\\")
```
